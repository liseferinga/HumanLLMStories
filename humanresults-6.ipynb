```python
%%html
<!-- hide-cell -->
<style>
.jp-Cell:has(#hide-this) .jp-Cell-inputWrapper { display: none !important; }
.jp-MarkdownCell h1 {
  font-family: sans-serif;
  font-size: 35px;
  font-weight: 600;
  color: #111;
  margin: 2.5rem 0 0.75rem 0;
}
.jp-MarkdownCell h2 {
  font-family: sans-serif;
  font-size: 25px;
  font-weight: 600;
  color: #333;
  margin: 2rem 0 0.75rem 0;
}
</style>
<span id="hide-this" style="display:none;"></span>
<style>
  .prose { font-size:15px; color:#111; line-height:1.75; margin:0 0 1rem 0; }
  .stat-grid { display:grid; grid-template-columns:repeat(auto-fit,minmax(140px,1fr)); gap:10px; margin:1.25rem 0; }
  .stat-card { background:#f8f8f8; border-radius:8px; padding:14px 16px; }
  .stat-label { font-size:12px; color:#777; margin:0 0 4px; }
  .stat-val { font-size:22px; font-weight:500; color:#111; margin:0; }
  .stat-sub { font-size:12px; color:#999; margin:4px 0 0; }
  .excl-row { display:flex; justify-content:space-between; font-size:13px; padding:7px 0; border-bottom:1px solid #eee; color:#333; }
  .excl-row:last-child { border-bottom:none; }
  .excl-n { font-weight:500; color:#777; }
  .kw-grid { display:grid; grid-template-columns:1fr 1fr; gap:10px; margin:1rem 0; }
  .kw-block { background:#f8f8f8; border-radius:8px; padding:12px 14px; }
  .kw-title { font-size:12px; font-weight:500; color:#777; margin:0 0 8px; text-transform:uppercase; letter-spacing:.05em; }
  .kw-list { margin:0; padding:0; list-style:none; }
  .kw-list li { font-size:13px; color:#333; padding:3px 0; border-bottom:1px solid #eee; }
  .kw-list li:last-child { border-bottom:none; }
  .design-card { background:#fff; border:1px solid #e5e5e5; border-radius:10px; padding:1rem 1.25rem; margin:1rem 0; }
  .design-label { font-size:12px; font-weight:500; color:#777; text-transform:uppercase; letter-spacing:.05em; margin:0 0 6px; }
  h3.section { font-size:15px; font-weight:500; color:#111; margin:1.75rem 0 0.5rem; border-bottom:1px solid #eee; padding-bottom:6px; }
</style>

<div style="overflow: visible; height: auto; max-height: none;">

<p class="prose">No standardized test or rubric can substitute for the direct evaluation of a creative product by knowledgeable raters (Baer &amp; McKool, 2014; Amabile, 1982)<sup><a href="#fn1">*</a></sup>.</p>

<div id="fn1" style="font-size:12px; color:#666;"><sup>*</sup>The TTCW framework does not replace human assessment but rather structures it and provides raters with explicit criteria defined by literary experts, while the ultimate evaluative judgment itself rests with the human reader (Chakrabarty et al., 2024). <a href="#fn1-ref">↩</a></div>

<h3 class="section">Recruitment</h3>

<p class="prose">In the current study, human assessors were recruited through a combination of convenience sampling via personal networks and social media dissemination. The call for participants was (re)posted on LinkedIn, posted on several Dutch writing blogs and Goodreads pages, and sent directly via e-mail to language and media studies professors. Participation was voluntary and unpaid.</p>

<h3 class="section">Sample</h3>

<div class="stat-grid">
  <div class="stat-card">
    <p class="stat-label">Final sample</p>
    <p class="stat-val">113</p>
    <p class="stat-sub">Fully complete responses</p>
  </div>
  <div class="stat-card">
    <p class="stat-label">Experts</p>
    <p class="stat-val">33</p>
  </div>
  <div class="stat-card">
    <p class="stat-label">Non-experts</p>
    <p class="stat-val">80</p>
  </div>
</div>

<div class="design-card">
  <p class="design-label">Exclusion breakdown</p>
  <div class="excl-row"><span>Raw questionnaire responses</span><span class="excl-n">134</span></div>
  <div class="excl-row"><span>No story ratings completed</span><span class="excl-n">7</span></div>
  <div class="excl-row"><span>Invariant responding only</span><span class="excl-n">2</span></div>
  <div class="excl-row"><span>Incomplete responses (missing ratings or attribution)</span><span class="excl-n">12</span></div>
  <div class="excl-row"><strong><span>Final analysis sample</span><span class="excl-n">113</span></strong></div>
</div>

<h3 class="section">Expertise classification</h3>

<p class="prose">Assessors ranged in age and background. Each provided basic demographic information prior to engaging with any story material, including age, gender, highest level of education completed, and field of study or occupation. This information was used to classify assessors as either expert or non-expert with respect to creative writing evaluation. Kaufman et al. (2013) demonstrates empirically that in the domain of short story evaluation specifically, quasi-experts (individuals with relevant domain experience but without full professional standing) correlate strongly with expert raters (r = .89) and can serve as valid substitutes, whereas novice raters show considerably lower agreement with experts (r = .56 on average with small samples). Chakrabarty et al. (2024) relied exclusively on expert raters, but explicitly called for future work to examine whether non-experts correlate with experts when performing TTCW evaluation. The present study answers this call. Here, expertise is treated as a fixed variable rather than an inclusion criterion, allowing for empirical examination of whether expert and non-expert assessors diverge in their TTCW-based evaluations of human versus LLM-generated stories.</p>

<p class="prose">Assessors were classified as expert if their educational background or professional occupation fell within language, literature, creative writing, linguistics, translation, journalism, or language teaching at an advanced level (WO Master or above, or equivalent professional experience, such as published author). The final sample comprised 33 expert and 80 non-expert assessors.</p>

<div class="kw-grid">
  <div class="kw-block">
    <p class="kw-title">Classified as expert</p>
    <ul class="kw-list">
      <li>Nederlandse taal- en letterkunde / Neerlandistiek</li>
      <li>Taal- en Letterkunde / Letterkunde</li>
      <li>Taalwetenschap / Taalkunde</li>
      <li>Vertaalwetenschap / Vertalen / Lexicografie</li>
      <li>Literatuurwetenschap</li>
      <li>Woordkunst</li>
      <li>Journalistiek</li>
      <li>Uitgever</li>
      <li>Docent Nederlands</li>
      <li>Historische letterkunde</li>
      <li>Vroegmoderne literatuur</li>
      <li>Toegepaste taalkunde</li>
      <li>Engelse / Franse taal en cultuur</li>
    </ul>
  </div>
</div>

<h3 class="section">Study design</h3>

<p class="prose">The questionnaire was administered in Dutch via Qualtrics. Each assessor was presented with two short stories drawn from the same prompt set: one written by a human author and one generated by a single LLM (either GPT, Gemini, or Claude). The assignment of the prompt set and the order logic was determined by the survey distribution logic of Qualtrics.</p>

<p class="prose">Initially, assessors were not aware that one of the two stories they were about to read was human-authored and the other generated by a language model. The study was framed as an investigation into how people evaluate creativity in short Dutch-language stories from writing competitions. This design choice was made deliberately to prevent assessors from searching for surface-level AI artefacts rather than responding to the stories as genuine pieces of writing.</p>

<div class="design-card">
  <p class="design-label">Rating procedure</p>
  <p class="prose" style="font-size:14px;margin:0;">After reading each story, assessors rated it on the 14 binary items of the Torrance Test for Creative Writing (Chakrabarty et al., 2024), responding <em>ja</em> (yes) or <em>nee</em> (no) to each. Each item was preceded by a brief explanation of the relevant construct, presented immediately before the rating items for that story, to support assessors with varying levels of literary expertise. The 14 items were identical for both stories. Finally, assessors were asked to indicate which of the two stories they believed had been generated by AI. Only assessors who completed all 14 items for both stories and provided an attribution answer were included in the final sample (n = 113).</p>
</div>

<p class="prose">Chakrabarty et al. (2024) report the passing rate of both human and LLM-written stories as the percentage of ratings answered with 'yes', broken down by author. This approach is followed in the current study, extended with inferential testing via binomial GLMMs.</p>

</div>
```

```python
def load_raw(path):
    df = pd.read_csv(path, skiprows=[1, 2])
    df.columns = (
        df.columns
        .str.replace("\xa0", " ", regex=False)
        .str.strip()
    )
    drop_cols = [
        "StartDate", "EndDate", "Status", "IPAddress", "Progress",
        "RecipientLastName", "RecipientFirstName", "RecipientEmail",
        "ExternalReference", "LocationLatitude", "LocationLongitude",
        "DistributionChannel", "UserLanguage", "Q5_4_TEXT", "StoryA", "StoryB",
    ]
    df = df.drop(columns=[c for c in drop_cols if c in df.columns])

    # llm_model wasn't captured by Qualtrics for these three responses;
    # confirmed against source records.
    MANUAL_LLM_MODEL = {
        "R_2MKmlAlIezyJ2ps": "claude",
        "R_2exxZTGKMM6B3oE": "claude",
        "R_8arEEH1cPJQGz3X": "gpt",
    }
    for rid, model in MANUAL_LLM_MODEL.items():
        if rid not in df["ResponseId"].values:
            warnings.warn(f"Manual llm_model correction skipped — ResponseId not found: {rid}")
            continue
        df.loc[df["ResponseId"] == rid, "llm_model"] = model

    # Q339 (attribution) wasn't captured by Qualtrics for these three responses; recovered from source records.
    MANUAL_ATTRIBUTION = {
        "R_2MKmlAlIezyJ2ps": "Verhaal 2",
        "R_2exxZTGKMM6B3oE": "Verhaal 2",
        "R_8arEEH1cPJQGz3X": "Verhaal 1",
    }
    for rid, answer in MANUAL_ATTRIBUTION.items():
        if rid not in df["ResponseId"].values:
            warnings.warn(f"Manual attribution correction skipped — ResponseId not found: {rid}")
            continue
        df.loc[df["ResponseId"] == rid, "Q339"] = answer

    # llm_file / human_file weren't captured by Qualtrics for these  responses; filenames confirmed against source records.
    MANUAL_LLM_FILE = {
        "R_2MKmlAlIezyJ2ps": "claude_story_11.txt",
        "R_2exxZTGKMM6B3oE": "claude_story_14.txt",
        "R_8arEEH1cPJQGz3X": "gpt_story_17.txt",
    }
    MANUAL_HUMAN_FILE = {
        "R_2MKmlAlIezyJ2ps": "human_story_11.txt",
        "R_2exxZTGKMM6B3oE": "human_story_14.txt",
        "R_8arEEH1cPJQGz3X": "human_story_17.txt",
    }
    for rid, fname in MANUAL_LLM_FILE.items():
        if rid not in df["ResponseId"].values:
            warnings.warn(f"Manual llm_file correction skipped — ResponseId not found: {rid}")
            continue
        df.loc[df["ResponseId"] == rid, "llm_file"] = fname

    for rid, fname in MANUAL_HUMAN_FILE.items():
        if rid not in df["ResponseId"].values:
            warnings.warn(f"Manual human_file correction skipped — ResponseId not found: {rid}")
            continue
        df.loc[df["ResponseId"] == rid, "human_file"] = fname

    return df


df_raw = load_raw(RAW_PATH)
print(f"Loaded {df_raw.shape[0]} rows, {df_raw.shape[1]} columns")
df_raw.head(4)
```

```python
def recode_ja_nee(df, cols):
    mapping = {"Ja": 1, "Nee": 0}
    valid_targets = {0, 1, 0.0, 1.0}

    for col in cols:
        values = set(df[col].dropna().unique())

        if values <= valid_targets:
            # already recoded (e.g. cell re-run)
            continue

        unexpected = values - set(mapping.keys())
        if unexpected:
            warnings.warn(f"Unexpected values in {col}: {unexpected}")

        df[col] = df[col].map(mapping)

    return df

df = recode_ja_nee(df, s1_cols + s2_cols)
```

```python
SUBSCALES = {
    "Fluency":     ["Fluency1", "Fluency2", "Fluency3", "Fluency4", "Fluency5"],
    "Flexibility": ["Flexibility1", "Flexibility2", "Flexibility3"],
    "Originality": ["Originality1", "Originality2", "Originality3"],
    "Elaboration": ["Elaboration1", "Elaboration2", "Elaboration3"],
}

for prefix in ["s1", "s2"]:
    for subscale, items in SUBSCALES.items():
        cols = [f"{prefix}_{item}" for item in items]
        df[f"{prefix}_{subscale}_score"] = df[cols].sum(axis=1, skipna=True)
        df[f"{prefix}_{subscale}_n_answered"] = df[cols].notna().sum(axis=1)

    all_item_cols = [f"{prefix}_{item}" for item in TTCW_ITEMS]
    df[f"{prefix}_total_score"] = df[all_item_cols].sum(axis=1, skipna=True)
    df[f"{prefix}_n_answered"] = df[all_item_cols].notna().sum(axis=1)
```

```python
def pick_by_position(df, position_col, s1_col, s2_col):
    return np.select(
        [df[position_col] == "s1", df[position_col] == "s2"],
        [df[s1_col], df[s2_col]],
        default=np.nan,
    )

# Dimension-level scores 
score_keys = list(SUBSCALES.keys()) + ["total"]
for key in score_keys:
    s1_col = f"s1_{key}_score" if key != "total" else "s1_total_score"
    s2_col = f"s2_{key}_score" if key != "total" else "s2_total_score"
    df[f"human_{key}_score"] = pick_by_position(df, "human_position", s1_col, s2_col)
    df[f"llm_{key}_score"] = pick_by_position(df, "llm_position", s1_col, s2_col)

# Item-level pass/fail (for the GLMM)
for item in TTCW_ITEMS:
    s1_col = f"s1_{item}"
    s2_col = f"s2_{item}"
    df[f"human_{item}"] = pick_by_position(df, "human_position", s1_col, s2_col)
    df[f"llm_{item}"] = pick_by_position(df, "llm_position", s1_col, s2_col)
```

```python
import pandas as pd
import numpy as np
import krippendorff
from IPython.display import display as ipy_display

# define questions 
DIMENSION_MAP = {
    1: "Fluency1", 2: "Fluency2", 3: "Fluency3", 4: "Fluency4", 5: "Fluency5",
    6: "Flexibility1", 7: "Flexibility2", 8: "Flexibility3",
    9: "Originality1", 10: "Originality2", 11: "Originality3",
    12: "Elaboration1", 13: "Elaboration2", 14: "Elaboration3",
}
model_data["dimension"] = model_data["question_id"].map(DIMENSION_MAP)

# story file
model_data = model_data.drop(columns=["llm_file", "human_file", "story_file"], errors="ignore")
file_map = dataset_B[["responseID", "llm_file", "human_file"]].rename(columns={"responseID": "rater_id"})
model_data = model_data.merge(file_map, on="rater_id", how="left")
model_data["story_file"] = np.where(
    model_data["source"] == "LLM",
    model_data["llm_file"],
    model_data["human_file"]
)

# Krippendorff's α (human stories only)
def compute_alpha(df_subset):
    try:
        matrix = df_subset.pivot_table(
            index="rater_id", columns="story_file", values="pass", aggfunc="mean"
        ).to_numpy()
        return round(krippendorff.alpha(reliability_data=matrix, level_of_measurement="nominal"), 3)
    except Exception:
        return np.nan

human_subset = model_data[model_data["source"] == "Human"]

alpha_human_overall   = compute_alpha(human_subset)
alpha_human_expert    = compute_alpha(human_subset[human_subset["expertise"] == "expert"])
alpha_human_nonexpert = compute_alpha(human_subset[human_subset["expertise"] == "non-expert"])

item_rows = []
for q_id in range(1, 15):
    item_subset = human_subset[human_subset["question_id"] == q_id]
    item_rows.append({
        "Item":           q_id,
        "Dimension":      DIMENSION_MAP[q_id],
        "α (Overall)":    compute_alpha(item_subset),
        "α (Expert)":     compute_alpha(item_subset[item_subset["expertise"] == "expert"]),
        "α (Non-expert)": compute_alpha(item_subset[item_subset["expertise"] == "non-expert"]),
    })

item_alpha_df = pd.DataFrame(item_rows)

alpha_summary = pd.DataFrame([{
    "Level":          "All 14 items (pooled)",
    "α (Overall)":    alpha_human_overall,
    "α (Expert)":     alpha_human_expert,
    "α (Non-expert)": alpha_human_nonexpert,
}])

# total scores 
total_long = (
    model_data.groupby(["rater_id", "expertise", "source"])["pass"]
    .sum().reset_index().rename(columns={"pass": "score"})
)

# descriptive stats: total TTCW by expertise + source 
def summarise(group_df, expertise_label, alpha_val):
    stats = (
        group_df.groupby(["source"], observed=True)["score"]
        .agg(N="count", Mean="mean", Median="median", SD="std", Min="min", Max="max")
        .reset_index()
    )
    stats["Expertise"] = expertise_label
    stats["Krippendorff's α"] = stats["source"].apply(
        lambda s: alpha_val if s == "Human" else np.nan
    )
    return stats

expertise_order = ["Total", "Expert raters", "Non-expert raters"]
source_order    = ["Human", "LLM"]

score_groups = [
    (total_long,                                          "Total",             alpha_human_overall),
    (total_long[total_long["expertise"] == "expert"],     "Expert raters",     alpha_human_expert),
    (total_long[total_long["expertise"] == "non-expert"], "Non-expert raters", alpha_human_nonexpert),
]

stats = pd.concat([summarise(df, label, alpha) for df, label, alpha in score_groups], ignore_index=True)
stats["Expertise"] = pd.Categorical(stats["Expertise"], categories=expertise_order, ordered=True)
stats["source"]    = pd.Categorical(stats["source"],    categories=source_order,    ordered=True)
stats = stats.sort_values(["Expertise", "source"]).reset_index(drop=True)
stats[["Mean", "Median", "SD", "Min", "Max"]] = stats[["Mean", "Median", "SD", "Min", "Max"]].round(2)

display_df = stats.rename(columns={"source": "Source"})[
    ["Expertise", "Source", "N", "Mean", "Median", "SD", "Min", "Max", "Krippendorff's α"]
]

print("Total TTCW scores (0–14) by expertise group and story source.")
ipy_display(
    display_df.style
    .hide(axis="index")
    .format({
        "Mean": "{:.2f}", "Median": "{:.2f}", "SD": "{:.2f}",
        "Min": "{:.2f}", "Max": "{:.2f}",
        "Krippendorff's α": lambda x: f"{x:.3f}" if pd.notna(x) else "—",
    })
    .set_caption(
        "Total TTCW scores (0–14) by expertise group and story source. "
        "Krippendorff's α reported for human stories only (mean 3.1 raters/story); "
        "LLM stories lack sufficient rater overlap for reliable estimation."
    )
    .set_table_styles([
        {"selector": "thead th", "props": [("text-align", "left")]},
        {"selector": "caption",  "props": [("text-align", "left")]},
        {"selector": "td",       "props": [("text-align", "left")]},
    ])
)

# descriptive stats: LLM scores by model + expertise 
llm_model_map  = dataset_B[["responseID", "llm_model"]].rename(columns={"responseID": "rater_id"})
total_long_llm = total_long[total_long["source"] == "LLM"].merge(llm_model_map, on="rater_id", how="left")

def summarise_llm(group_df, expertise_label):
    stats = (
        group_df.groupby(["llm_model"], observed=True)["score"]
        .agg(N="count", Mean="mean", Median="median", SD="std", Min="min", Max="max")
        .reset_index()
    )
    stats["Expertise"] = expertise_label
    return stats

llm_groups = [
    (total_long_llm,                                              "Total"),
    (total_long_llm[total_long_llm["expertise"] == "expert"],     "Expert raters"),
    (total_long_llm[total_long_llm["expertise"] == "non-expert"], "Non-expert raters"),
]

llm_stats = pd.concat([summarise_llm(df, label) for df, label in llm_groups], ignore_index=True)
llm_stats[["Mean", "Median", "SD", "Min", "Max"]] = llm_stats[["Mean", "Median", "SD", "Min", "Max"]].round(2)

llm_stats["llm_model"] = pd.Categorical(llm_stats["llm_model"], categories=["claude", "gemini", "gpt"], ordered=True)
llm_stats["Expertise"] = pd.Categorical(llm_stats["Expertise"], categories=expertise_order, ordered=True)
llm_stats = llm_stats.sort_values(["Expertise", "llm_model"]).reset_index(drop=True)

display_llm = llm_stats.rename(columns={"llm_model": "LLM Model"})[
    ["Expertise", "LLM Model", "N", "Mean", "Median", "SD", "Min", "Max"]
]

# expert rater coverage per LLM model
llm_model_map_exp = dataset_B[["responseID", "llm_model"]].rename(columns={"responseID": "rater_id"})
expert_llm = model_data[
    (model_data["expertise"] == "expert") & (model_data["source"] == "LLM")
].merge(llm_model_map_exp, on="rater_id", how="left")

coverage = (
    expert_llm.groupby("llm_model")["rater_id"]
    .nunique()
    .rename("N expert raters")
    .reset_index()
    .rename(columns={"llm_model": "LLM Model"})
)
coverage["LLM Model"] = pd.Categorical(coverage["LLM Model"], categories=["claude", "gemini", "gpt"], ordered=True)
coverage = coverage.sort_values("LLM Model").reset_index(drop=True)

print("\nExpert rater coverage per LLM model (used to check balance before interpreting model differences).")
ipy_display(
    coverage.style
    .hide(axis="index")
    .set_table_styles([
        {"selector": "thead th", "props": [("text-align", "left")]},
        {"selector": "td",       "props": [("text-align", "left")]},
    ])
)

print("\nTotal TTCW scores (0–14) for LLM stories by model and expertise group.")
ipy_display(
    display_llm.style
    .hide(axis="index")
    .format({"Mean": "{:.2f}", "Median": "{:.2f}", "SD": "{:.2f}", "Min": "{:.2f}", "Max": "{:.2f}"})
    .set_caption(
        "Total TTCW scores (0–14) for LLM stories only, broken down by LLM model "
        "(Claude, Gemini, GPT) and expertise group."
    )
    .set_table_styles([
        {"selector": "thead th", "props": [("text-align", "left")]},
        {"selector": "caption",  "props": [("text-align", "left")]},
        {"selector": "td",       "props": [("text-align", "left")]},
    ])
)

#  Krippendorff's α summary
print("\nKrippendorff's α for all 14 TTCW items per expertise group (human stories only)")
ipy_display(
    alpha_summary.style.hide(axis="index")
    .format({"α (Overall)": "{:.3f}", "α (Expert)": "{:.3f}", "α (Non-expert)": "{:.3f}"})
)


print("\nKrippendorff's α per TTCW item (human stories only)")
ipy_display(
    item_alpha_df.style.hide(axis="index")
    .format({"α (Overall)": "{:.3f}", "α (Expert)": "{:.3f}", "α (Non-expert)": "{:.3f}"})
)
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import cohen_kappa_score
from IPython.display import display as ipy_display

# ── Majority vote helper ───────────────────────────────────────────────────────
def majority_vote(values):
    counts = values.value_counts()
    if len(counts) == 2 and counts.iloc[0] == counts.iloc[1]:
        return np.nan
    return counts.idxmax()

# ── Consensus kappa helper ────────────────────────────────────────────────────
def compute_consensus_kappa(data, item_id_col):
    rows = []
    for (item_id, qid), group in data.groupby([item_id_col, "question_id"]):
        experts    = group.loc[group["expertise"] == "expert",     "pass"]
        nonexperts = group.loc[group["expertise"] == "non-expert", "pass"]
        if experts.empty or nonexperts.empty:
            continue
        exp_consensus = majority_vote(experts)
        ne_consensus  = majority_vote(nonexperts)
        if pd.isna(exp_consensus) or pd.isna(ne_consensus):
            continue
        rows.append({
            "item_id":             item_id,
            "question_id":         qid,
            "expert_consensus":    exp_consensus,
            "nonexpert_consensus": ne_consensus,
            "n_experts":           len(experts),
            "n_nonexperts":        len(nonexperts),
        })
    result    = pd.DataFrame(rows)
    n_items   = len(result)
    pct_agree = (result["expert_consensus"] == result["nonexpert_consensus"]).mean() * 100
    kappa     = cohen_kappa_score(result["expert_consensus"], result["nonexpert_consensus"])
    return n_items, pct_agree, kappa, result

# ── Human stories ─────────────────────────────────────────────────────────────
human_data            = model_data[model_data["source"] == "Human"].copy()
human_data["file_id"] = "set" + human_data["set"].astype(str)
n_human, agree_human, kappa_human, details_human = compute_consensus_kappa(human_data, "file_id")

# ── LLM stories ───────────────────────────────────────────────────────────────
llm_data            = model_data[model_data["source"] == "LLM"].copy()
llm_data["file_id"] = "set" + llm_data["set"].astype(str) + "_" + llm_data["llm_author"].astype(str)
n_llm, agree_llm, kappa_llm, details_llm = compute_consensus_kappa(llm_data, "file_id")

# ── Summary table ─────────────────────────────────────────────────────────────
kappa_summary = pd.DataFrame([
    {
        "Story type":        "Human",
        "N item comparisons": n_human,
        "Agreement (%)":     round(agree_human, 1),
        "Cohen's κ":         round(kappa_human, 3),
    },
    {
        "Story type":        "LLM",
        "N item comparisons": n_llm,
        "Agreement (%)":     round(agree_llm, 1),
        "Cohen's κ":         round(kappa_llm, 3),
    },
])

ipy_display(
    kappa_summary.style
    .hide(axis="index")
    .format({"Agreement (%)": "{:.1f}", "Cohen's κ": "{:.3f}"})
    .set_caption(
        "Expert vs. non-expert consensus agreement (majority vote per story × item). "
        "N = number of story–item pairs where both groups had a non-tied majority. "
        "Cohen's κ is corrected for chance agreement."
    )
    .set_table_styles([
        {"selector": "thead th", "props": [("text-align", "left")]},
        {"selector": "caption",  "props": [("text-align", "left")]},
        {"selector": "td",       "props": [("text-align", "left")]},
    ])
)
```

```python
from scipy.stats import binomtest
from IPython.display import display as ipy_display
import pandas as pd

def fmt_p(p):
    if p < .001:
        return "< .001 ***"
    elif p < .01:
        return f"= {p:.3f} **"
    elif p < .05:
        return f"= {p:.3f} *"
    else:
        return f"= {p:.3f}"

def add_binom(df, n_col="N", correct_col="n_correct"):
    ps = []
    for _, row in df.iterrows():
        result = binomtest(int(row[correct_col]), int(row[n_col]), p=0.5, alternative="two-sided")
        ps.append(fmt_p(result.pvalue))
    df = df.copy()
    df["p (vs. chance)"] = ps
    return df

def fmt_accuracy(df):
    df = df.copy()
    df["Accuracy (%)"] = df.apply(
        lambda row: f"{row['Accuracy (%)']:.1f} (n={row['n_correct']}/{row['N']})", axis=1
    )
    return df

CAPTION_SUFFIX = "Chance level = 50%. * p<.05, ** p<.01, *** p<.001."
TABLE_STYLES = [
    {"selector": "thead th", "props": [("text-align", "left")]},
    {"selector": "caption",  "props": [("text-align", "left")]},
    {"selector": "td",       "props": [("text-align", "left")]},
]

# ── Overall ───────────────────────────────────────────────────────────────────
total_acc = dataset_B["attribution_correct"].agg(["count", "sum", "mean"])
n_total   = int(total_acc["count"])
k_total   = int(total_acc["sum"])
p_total   = binomtest(k_total, n_total, p=0.5, alternative="two-sided").pvalue
print(
    f"Overall attribution accuracy: "
    f"{total_acc['mean']*100:.1f}% "
    f"(n={k_total}/{n_total}), p {fmt_p(p_total)}"
)

# ── By expertise ──────────────────────────────────────────────────────────────
acc_exp = (
    dataset_B
    .groupby("expertise", observed=True)["attribution_correct"]
    .agg(N="count", n_correct="sum", **{"Accuracy (%)": "mean"})
    .reset_index()
)
acc_exp["Accuracy (%)"] = (acc_exp["Accuracy (%)"] * 100).round(1)
acc_exp["n_correct"]    = acc_exp["n_correct"].astype(int)
acc_exp["expertise"]    = pd.Categorical(
    acc_exp["expertise"], categories=["expert", "non-expert"], ordered=True
)
acc_exp = acc_exp.sort_values("expertise").reset_index(drop=True)
acc_exp = add_binom(acc_exp)
acc_exp = fmt_accuracy(acc_exp)

display_exp = acc_exp.rename(columns={"expertise": "Expertise"})
display_exp["Expertise"] = display_exp["Expertise"].cat.rename_categories({
    "expert": "Expert raters", "non-expert": "Non-expert raters",
})
display_exp = display_exp[["Expertise", "N", "Accuracy (%)", "p (vs. chance)"]]

print("\nAttribution accuracy by expertise group.")
ipy_display(
    display_exp.style
    .hide(axis="index")
    .set_caption(
        "Attribution accuracy by expertise group. "
        "N = number of respondents. "
        "Accuracy (%) = proportion who correctly identified the AI-written story. "
        + CAPTION_SUFFIX
    )
    .set_table_styles(TABLE_STYLES)
)

# ── By expertise × LLM model ──────────────────────────────────────────────────
acc_model = (
    dataset_B
    .groupby(["expertise", "llm_model"], observed=True)["attribution_correct"]
    .agg(N="count", n_correct="sum", **{"Accuracy (%)": "mean"})
    .reset_index()
)

acc_model_total = (
    dataset_B
    .groupby("llm_model", observed=True)["attribution_correct"]
    .agg(N="count", n_correct="sum", **{"Accuracy (%)": "mean"})
    .reset_index()
)
acc_model_total["expertise"] = "total"

acc_model = pd.concat([acc_model, acc_model_total], ignore_index=True)
acc_model["Accuracy (%)"] = (acc_model["Accuracy (%)"] * 100).round(1)
acc_model["n_correct"]    = acc_model["n_correct"].astype(int)
acc_model = add_binom(acc_model)
acc_model = fmt_accuracy(acc_model)

acc_model["expertise"] = pd.Categorical(
    acc_model["expertise"], categories=["total", "expert", "non-expert"], ordered=True
)
acc_model["llm_model"] = pd.Categorical(
    acc_model["llm_model"], categories=["claude", "gemini", "gpt"], ordered=True
)
acc_model = acc_model.sort_values(["llm_model", "expertise"]).reset_index(drop=True)

display_model = acc_model.rename(columns={
    "expertise": "Expertise", "llm_model": "LLM Model"
})[["LLM Model", "Expertise", "N", "Accuracy (%)", "p (vs. chance)"]]
display_model["Expertise"] = display_model["Expertise"].cat.rename_categories({
    "total": "Total", "expert": "Expert raters", "non-expert": "Non-expert raters",
})

def style_total(row):
    return ["font-weight: bold" if row["Expertise"] == "Total" else "" for _ in row.index]

print("\nAttribution accuracy by LLM model and expertise group.")
ipy_display(
    display_model.style
    .apply(style_total, axis=1)
    .hide(axis="index")
    .set_caption(
        "Attribution accuracy by LLM model and expertise group. "
        "N = number of respondents who evaluated that model. "
        "Total row = pooled across expert and non-expert raters. "
        + CAPTION_SUFFIX
    )
    .set_table_styles(TABLE_STYLES)
)
```

```python
import pandas as pd
import numpy as np
import warnings

DIMENSION_MAP = {
    1: "Fluency1", 2: "Fluency2", 3: "Fluency3", 4: "Fluency4", 5: "Fluency5",
    6: "Flexibility1", 7: "Flexibility2", 8: "Flexibility3",
    9: "Originality1", 10: "Originality2", 11: "Originality3",
    12: "Elaboration1", 13: "Elaboration2", 14: "Elaboration3",
}

rows = []
for _, row in dataset_B.iterrows():
    for item in TTCW_ITEMS:
        q_id = ITEM_TO_QID[item]
        for source, col_prefix, file_col in [
            ("Human", "human", "human_file"),
            ("LLM",   "llm",   "llm_file"),
        ]:
            col = f"{col_prefix}_{item}"
            if col not in dataset_B.columns:
                continue
            rows.append({
                "rater_id":    row["responseID"],
                "expertise":   row["expertise"],
                "source":      source,
                "question_id": q_id,
                "pass":        row[col],
                "set":         row["set"],
                "order":       row["order"],
                "llm_author":  "Human" if source == "Human" else row["llm_model"],
                "story_file":  row[file_col],
            })

model_data = pd.DataFrame(rows)
model_data = model_data.dropna(subset=["pass"])
model_data["pass"]      = model_data["pass"].astype(int)
model_data["order"]     = model_data["order"].map({"llm_first": "LLM_first", "human_first": "Human_first"})
model_data["obs_id"]    = range(len(model_data))
model_data["dimension"] = model_data["question_id"].map(DIMENSION_MAP)

# ── Dummy coding (Claude = reference) ────────────────────────────────────────
model_data["author_gpt_vs_claude"]    = (model_data["llm_author"].str.lower() == "gpt").astype(int)
model_data["author_gemini_vs_claude"] = (model_data["llm_author"].str.lower() == "gemini").astype(int)

# ── Story file identifiers ────────────────────────────────────────────────────
file_map = dataset_B[["responseID", "llm_file", "human_file"]].rename(columns={"responseID": "rater_id"})
model_data = model_data.merge(file_map, on="rater_id", how="left")

# ── Merge contest quality ─────────────────────────────────────────────────────
CONTEST_QUALITY_PATH = "Scoring Model Dutch Writing Contests - Scoring.csv"
contest_quality = pd.read_csv(CONTEST_QUALITY_PATH)
contest_quality = contest_quality.rename(columns={"ID": "set", "End Score": "end_score"})
contest_quality = contest_quality[["set", "end_score"]].copy()
model_data["set"]      = pd.to_numeric(model_data["set"],      errors="coerce")
contest_quality["set"] = pd.to_numeric(contest_quality["set"], errors="coerce")
model_data = model_data.merge(contest_quality, on="set", how="left")

missing = model_data["end_score"].isna().sum()
if missing:
    warnings.warn(f"{missing} rows had no matching End Score")
else:
    print(f"End Score matched for all {len(model_data)} rows")

model_data["end_score_c"] = model_data["end_score"] - model_data["end_score"].mean()

# ── model_df alias (used by attribution cell) ─────────────────────────────────
model_df = model_data.copy()

# ── Export ────────────────────────────────────────────────────────────────────
model_data.to_csv("model_data.csv", index=False)
print(f"model_data shape: {model_data.shape}")
print("\nLLM author value counts:")
print(model_data.loc[model_data["source"] == "LLM", "llm_author"].value_counts())
print("\nDummy code sums:")
print(model_data[["author_gpt_vs_claude", "author_gemini_vs_claude"]].sum())
model_data.head()
```

```python
r_script = '''
# analysis_glmm_unified.R
required_packages <- c("lme4", "performance", "DHARMa", "broom.mixed", "emmeans")
missing_packages <- required_packages[!required_packages %in% installed.packages()[, "Package"]]
if (length(missing_packages) > 0) {
  install.packages(missing_packages, repos = "https://cloud.r-project.org")
}
suppressMessages({
  library(lme4)
  library(performance)
  library(DHARMa)
  library(broom.mixed)
  library(emmeans)
})

set.seed(42)

args <- commandArgs(trailingOnly = TRUE)
if (length(args) < 11) stop(sprintf("Expected 11 arguments, got %d.", length(args)))

data_path                <- args[1]
model_path               <- args[2]
fixed_path               <- args[3]
diag_path                <- args[4]
omnibus_author_path      <- args[5]
omnibus_qid_path         <- args[6]
omnibus_interaction_path <- args[7]
vs_human_path            <- args[8]
pairwise_path            <- args[9]
vs_human_by_expertise_path <- args[10]
random_path              <- args[11]

df <- read.csv(data_path, stringsAsFactors = FALSE)
df$source      <- factor(df$source, levels = c("Human", "LLM"))
df$question_id <- factor(df$question_id)
df$order       <- factor(df$order, levels = c("Human_first", "LLM_first"))
df$expertise   <- factor(df$expertise, levels = c("expert", "non-expert"))
df$rater_id    <- factor(df$rater_id)
df$set         <- factor(df$set)
df$obs_id      <- factor(df$obs_id)
df$end_score_c <- as.numeric(df$end_score_c)

df$author <- with(df, ifelse(
  source == "Human", "Human",
  ifelse(author_gpt_vs_claude == 1, "GPT",
  ifelse(author_gemini_vs_claude == 1, "Gemini", "Claude"))
))
df$author <- factor(df$author, levels = c("Human", "GPT", "Gemini", "Claude"))

# ── BINARY MODEL ──────────────────────────────────────────────────────────────
cat("=== Binary model: LLM vs. Human (collapsed) ===\\n")
model_binary <- glmer(
  pass ~ source * expertise + question_id + order + end_score_c +
    (1 | rater_id) + (1 | set),
  data = df, family = binomial(link = "logit"),
  control = glmerControl(optimizer = "bobyqa", optCtrl = list(maxfun = 2e5))
)
print(summary(model_binary))

model_binary_no_source <- update(model_binary, . ~ . - source - source:expertise)
omnibus_binary_source <- anova(model_binary, model_binary_no_source)
write.csv(data.frame(
  term = "source (LLM vs Human, omnibus)",
  chisq = omnibus_binary_source$Chisq[2],
  df = omnibus_binary_source$"Df"[2],
  p_value = omnibus_binary_source$"Pr(>Chisq)"[2]
), "glmm_omnibus_binary_source.csv", row.names = FALSE)

model_binary_no_interaction <- update(model_binary, . ~ . - source:expertise)
omnibus_binary_interaction <- anova(model_binary, model_binary_no_interaction)
write.csv(data.frame(
  term = "source:expertise (omnibus)",
  chisq = omnibus_binary_interaction$Chisq[2],
  df = omnibus_binary_interaction$"Df"[2],
  p_value = omnibus_binary_interaction$"Pr(>Chisq)"[2]
), "glmm_omnibus_binary_interaction.csv", row.names = FALSE)

emm_source <- emmeans(model_binary, ~ source)
vs_human_binary <- contrast(emm_source, "trt.vs.ctrl", ref = "Human", adjust = "fdr")
vs_human_binary_df <- as.data.frame(summary(vs_human_binary, infer = c(TRUE, TRUE)))
write.csv(vs_human_binary_df, "glmm_llm_vs_human_binary.csv", row.names = FALSE)

emm_source_by_expertise <- emmeans(model_binary, ~ source | expertise)
vs_human_binary_by_exp <- contrast(emm_source_by_expertise, "trt.vs.ctrl", ref = "Human", adjust = "fdr")
vs_human_binary_by_exp_df <- as.data.frame(summary(vs_human_binary_by_exp, infer = c(TRUE, TRUE)))
write.csv(vs_human_binary_by_exp_df, "glmm_llm_vs_human_binary_by_expertise.csv", row.names = FALSE)

tidy_binary <- broom.mixed::tidy(model_binary, effects = "fixed", conf.int = TRUE, conf.method = "Wald")
tidy_binary$odds_ratio <- exp(tidy_binary$estimate)
write.csv(tidy_binary, "glmm_binary_fixed_effects.csv", row.names = FALSE)

# ── PRIMARY MODEL: author x expertise ─────────────────────────────────────────
model1 <- glmer(
  pass ~ author * expertise + question_id + order + end_score_c +
    (1 | rater_id) + (1 | set),
  data = df, family = binomial(link = "logit"),
  control = glmerControl(optimizer = "bobyqa", optCtrl = list(maxfun = 2e5))
)
cat("=== Model 1: binomial GLMM (author x expertise) ===\\n")
print(summary(model1))

sim_res <- simulateResiduals(model1, n = 1000)
disp_test <- testDispersion(sim_res, plot = FALSE)
cat("\\n=== Overdispersion test (DHARMa) ===\\n")
print(disp_test)

overdispersed <- disp_test$p.value < 0.05
final_model <- model1
model_used  <- "Model 1: binomial GLMM (no OLRE)"

if (overdispersed) {
  model2 <- glmer(
    pass ~ author * expertise + question_id + order + end_score_c +
      (1 | rater_id) + (1 | set) + (1 | obs_id),
    data = df, family = binomial(link = "logit"),
    control = glmerControl(optimizer = "bobyqa", optCtrl = list(maxfun = 2e5))
  )
  final_model <- model2
  model_used  <- "Model 2: binomial GLMM + OLRE (overdispersion correction)"
}

conv_warnings  <- final_model@optinfo$conv$lme4$messages
convergence_ok <- is.null(conv_warnings)
cat("\\n=== Convergence diagnostics ===\\n")
if (convergence_ok) cat("No convergence warnings.\\n") else print(conv_warnings)

cat("\\n=== Multicollinearity (VIF) ===\\n")
vif_result <- tryCatch(check_collinearity(final_model), error = function(e) NULL)
if (!is.null(vif_result)) print(vif_result)
max_vif <- if (!is.null(vif_result)) max(vif_result$VIF) else NA

# ── Omnibus tests ─────────────────────────────────────────────────────────────
model_no_author <- update(final_model, . ~ . - author - author:expertise)
omnibus_author_result <- anova(final_model, model_no_author)
write.csv(data.frame(
  term = "author (omnibus)",
  chisq = omnibus_author_result$Chisq[2],
  df = omnibus_author_result$"Df"[2],
  p_value = omnibus_author_result$"Pr(>Chisq)"[2]
), omnibus_author_path, row.names = FALSE)

model_no_qid <- update(final_model, . ~ . - question_id)
omnibus_qid_result <- anova(final_model, model_no_qid)
write.csv(data.frame(
  term = "question_id (omnibus)",
  chisq = omnibus_qid_result$Chisq[2],
  df = omnibus_qid_result$"Df"[2],
  p_value = omnibus_qid_result$"Pr(>Chisq)"[2]
), omnibus_qid_path, row.names = FALSE)

model_no_interaction <- update(final_model, . ~ . - author:expertise)
omnibus_interaction_result <- anova(final_model, model_no_interaction)
write.csv(data.frame(
  term = "author:expertise (omnibus)",
  chisq = omnibus_interaction_result$Chisq[2],
  df = omnibus_interaction_result$"Df"[2],
  p_value = omnibus_interaction_result$"Pr(>Chisq)"[2]
), omnibus_interaction_path, row.names = FALSE)

# ── emmeans contrasts ─────────────────────────────────────────────────────────
emm_author <- emmeans(final_model, ~ author)
vs_human <- contrast(emm_author, "trt.vs.ctrl", ref = "Human", adjust = "fdr")
write.csv(as.data.frame(summary(vs_human, infer = c(TRUE, TRUE))), vs_human_path, row.names = FALSE)

pairwise <- contrast(emm_author, "pairwise", adjust = "fdr")
write.csv(as.data.frame(summary(pairwise, infer = c(TRUE, TRUE))), pairwise_path, row.names = FALSE)

emm_author_by_expertise <- emmeans(final_model, ~ author | expertise)
vs_human_by_expertise <- contrast(emm_author_by_expertise, "trt.vs.ctrl", ref = "Human", adjust = "fdr")
write.csv(as.data.frame(summary(vs_human_by_expertise, infer = c(TRUE, TRUE))), vs_human_by_expertise_path, row.names = FALSE)

# ── Exports ───────────────────────────────────────────────────────────────────
tidy_fixed <- broom.mixed::tidy(final_model, effects = "fixed", conf.int = TRUE, conf.method = "Wald")
tidy_fixed$odds_ratio <- exp(tidy_fixed$estimate)
write.csv(tidy_fixed, fixed_path, row.names = FALSE)

write.csv(data.frame(
  model_used = model_used,
  overdispersion_detected = overdispersed,
  overdispersion_p_value = disp_test$p.value,
  convergence_ok = convergence_ok,
  max_vif = max_vif,
  vif_flag = ifelse(!is.na(max_vif) && max_vif > 5, "VIF > 5 (concern)", "OK"),
  seed_used = 42
), diag_path, row.names = FALSE)

random_fx <- broom.mixed::tidy(final_model, effects = "ran_pars", scales = "sdcor")
random_fx <- random_fx[grepl("^sd", random_fx$term), c("group", "estimate")]
names(random_fx)[names(random_fx) == "estimate"] <- "sd"
write.csv(random_fx, random_path, row.names = FALSE)

saveRDS(final_model, model_path)
writeLines(capture.output(sessionInfo()), "r_session_info.txt")
cat("\\nDone. Model saved to:", model_path, "\\n")
'''

with open("analysis_glmm_unified.R", "w") as f:
    f.write(r_script)

result = subprocess.run(
    [
        "Rscript", "analysis_glmm_unified.R", "model_data.csv", "glmm_model.rds",
        "glmm_fixed_effects.csv", "glmm_diagnostics.csv",
        "glmm_omnibus_author.csv", "glmm_omnibus_question_id.csv",
        "glmm_omnibus_interaction.csv",
        "glmm_author_vs_human.csv", "glmm_author_pairwise.csv",
        "glmm_author_vs_human_by_expertise.csv",
        "glmm_random_effects.csv",
    ],
    capture_output=True, text=True,
)
print(result.stdout)
print(result.stderr)
```

```python
r_script = '''
# analysis_expertise_interaction.R
required_packages <- c("lme4", "emmeans")
missing_packages <- required_packages[!required_packages %in% installed.packages()[, "Package"]]
if (length(missing_packages) > 0) {
  install.packages(missing_packages, repos = "https://cloud.r-project.org")
}
suppressMessages({ library(lme4); library(emmeans) })

set.seed(42)

args <- commandArgs(trailingOnly = TRUE)
data_path                <- args[1]
omnibus_interaction_path <- args[2]
simple_effects_path      <- args[3]

df <- read.csv(data_path, stringsAsFactors = FALSE)
df$source      <- factor(df$source, levels = c("Human", "LLM"))
df$question_id <- factor(df$question_id)
df$order       <- factor(df$order, levels = c("Human_first", "LLM_first"))
df$expertise   <- factor(df$expertise, levels = c("non-expert", "expert"))
df$rater_id    <- factor(df$rater_id)
df$set         <- factor(df$set)
if ("end_score_c" %in% names(df)) df$end_score_c <- as.numeric(df$end_score_c)

model_interaction <- glmer(
  pass ~ source + author_gpt_vs_claude + author_gemini_vs_claude +
    question_id * expertise + order + end_score_c +
    (1 | rater_id) + (1 | set),
  data = df, family = binomial(link = "logit"),
  control = glmerControl(optimizer = "bobyqa", optCtrl = list(maxfun = 2e5))
)
cat("=== Model with question_id x expertise interaction ===\\n")
print(summary(model_interaction))

conv_warnings <- model_interaction@optinfo$conv$lme4$messages
if (is.null(conv_warnings)) cat("\\nNo convergence warnings.\\n") else print(conv_warnings)

cat("\\n=== Omnibus test: question_id x expertise interaction ===\\n")
model_no_interaction <- update(model_interaction, . ~ . - question_id:expertise)
omnibus_result <- anova(model_interaction, model_no_interaction)
print(omnibus_result)

omnibus_df <- data.frame(
  term = "question_id:expertise (interaction)",
  chisq = omnibus_result$Chisq[2],
  df = omnibus_result$"Df"[2],
  p_value = omnibus_result$"Pr(>Chisq)"[2]
)
write.csv(omnibus_df, omnibus_interaction_path, row.names = FALSE)

if (omnibus_df$p_value >= 0.05) {
  cat("\\nNOTE: the omnibus interaction test is NOT significant.\\n")
}

cat("\\n=== Per-item simple effects (expert vs. non-expert) ===\\n")
emm <- emmeans(model_interaction, ~ expertise | question_id)
simple_contrasts <- contrast(emm, "revpairwise", by = "question_id", adjust = "none")
simple_df <- as.data.frame(simple_contrasts)
simple_df$p.adj.fdr <- p.adjust(simple_df$p.value, method = "fdr")
print(simple_df)
write.csv(simple_df, simple_effects_path, row.names = FALSE)
cat("\\nDone.\\n")
'''

with open("analysis_expertise_interaction.R", "w") as f:
    f.write(r_script)

result = subprocess.run(
    ["Rscript", "analysis_expertise_interaction.R", "model_data.csv",
     "glmm_expertise_interaction_omnibus.csv", "glmm_expertise_simple_effects.csv"],
    capture_output=True, text=True,
)
print(result.stdout)
print(result.stderr)
```

```python
r_script = '''
# analysis_source_interaction.R
required_packages <- c("lme4", "emmeans")
missing_packages <- required_packages[!required_packages %in% installed.packages()[, "Package"]]
if (length(missing_packages) > 0) {
  install.packages(missing_packages, repos = "https://cloud.r-project.org")
}
suppressMessages({ library(lme4); library(emmeans) })

set.seed(42)

args <- commandArgs(trailingOnly = TRUE)
data_path                <- args[1]
omnibus_interaction_path <- args[2]
simple_effects_path      <- args[3]

df <- read.csv(data_path, stringsAsFactors = FALSE)
df$source      <- factor(df$source, levels = c("Human", "LLM"))
df$question_id <- factor(df$question_id)
df$order       <- factor(df$order, levels = c("Human_first", "LLM_first"))
df$expertise   <- factor(df$expertise, levels = c("non-expert", "expert"))
df$rater_id    <- factor(df$rater_id)
df$set         <- factor(df$set)
if ("end_score_c" %in% names(df)) df$end_score_c <- as.numeric(df$end_score_c)

model_interaction <- glmer(
  pass ~ author_gpt_vs_claude + author_gemini_vs_claude +
    question_id * source + order + expertise + end_score_c +
    (1 | rater_id) + (1 | set),
  data = df, family = binomial(link = "logit"),
  control = glmerControl(optimizer = "bobyqa", optCtrl = list(maxfun = 2e5))
)
cat("=== Model with question_id x source interaction ===\\n")
print(summary(model_interaction))

conv_warnings <- model_interaction@optinfo$conv$lme4$messages
if (is.null(conv_warnings)) cat("\\nNo convergence warnings.\\n") else print(conv_warnings)

cat("\\n=== Omnibus test: question_id x source interaction ===\\n")
model_no_interaction <- update(model_interaction, . ~ . - question_id:source)
omnibus_result <- anova(model_interaction, model_no_interaction)
print(omnibus_result)

omnibus_df <- data.frame(
  term = "question_id:source (interaction)",
  chisq = omnibus_result$Chisq[2],
  df = omnibus_result$"Df"[2],
  p_value = omnibus_result$"Pr(>Chisq)"[2]
)
write.csv(omnibus_df, omnibus_interaction_path, row.names = FALSE)

if (omnibus_df$p_value >= 0.05) {
  cat("\\nNOTE: the omnibus interaction test is NOT significant.\\n")
}

cat("\\n=== Per-item simple effects (Human vs. LLM) ===\\n")
emm <- emmeans(model_interaction, ~ source | question_id)
simple_contrasts <- contrast(emm, "revpairwise", by = "question_id", adjust = "none")
simple_df <- as.data.frame(simple_contrasts)
simple_df$p.adj.fdr <- p.adjust(simple_df$p.value, method = "fdr")
print(simple_df)
write.csv(simple_df, simple_effects_path, row.names = FALSE)
cat("\\nDone.\\n")
'''

with open("analysis_source_interaction.R", "w") as f:
    f.write(r_script)

result = subprocess.run(
    ["Rscript", "analysis_source_interaction.R", "model_data.csv",
     "glmm_source_interaction_omnibus.csv", "glmm_source_simple_effects.csv"],
    capture_output=True, text=True,
)
print(result.stdout)
print(result.stderr)
```

```python
# ── Prepare attribution data ──────────────────────────────────────────────────
attribution_df = dataset_B[
    ["responseID", "expertise", "order", "set", "llm_model", "attribution_correct"]
].copy()
attribution_df = attribution_df.rename(columns={"attribution_correct": "correct"})
attribution_df["pair_id"] = (
    attribution_df["set"].astype(str) + "_" + attribution_df["llm_model"].str.lower()
)
attribution_df["author_gpt_vs_claude"]    = (attribution_df["llm_model"].str.lower() == "gpt").astype(int)
attribution_df["author_gemini_vs_claude"] = (attribution_df["llm_model"].str.lower() == "gemini").astype(int)
attribution_df["order"] = attribution_df["order"].map({"llm_first": "LLM_first", "human_first": "Human_first"})
attribution_df = attribution_df.rename(columns={"responseID": "rater_id"})
attribution_df = attribution_df.dropna(subset=["correct"])
attribution_df["correct"] = attribution_df["correct"].astype(int)

rater_totals = model_df.groupby(["rater_id", "source"])["pass"].sum().unstack("source")
rater_totals = rater_totals.rename(columns={"Human": "human_total", "LLM": "llm_total"})
rater_totals["score_diff"] = rater_totals["llm_total"] - rater_totals["human_total"]

attribution_df = attribution_df.merge(
    rater_totals[["human_total", "llm_total", "score_diff"]],
    left_on="rater_id", right_index=True, how="left",
)
attribution_df["score_diff_z"] = (
    (attribution_df["score_diff"] - attribution_df["score_diff"].mean())
    / attribution_df["score_diff"].std()
)
attribution_df = attribution_df.dropna(subset=["score_diff_z"])
attribution_df.to_csv("attribution_data.csv", index=False)
print(f"Attribution data: {len(attribution_df)} rows")

# ── Write R script ────────────────────────────────────────────────────────────
r_script = '''
# analysis_attribution.R
required_packages <- c("lme4", "performance", "DHARMa", "broom.mixed")
missing_packages <- required_packages[!required_packages %in% installed.packages()[, "Package"]]
if (length(missing_packages) > 0) {
  install.packages(missing_packages, repos = "https://cloud.r-project.org")
}
suppressMessages({
  library(lme4); library(performance); library(DHARMa); library(broom.mixed)
})

set.seed(42)

args         <- commandArgs(trailingOnly = TRUE)
data_path    <- args[1]
fixed_path   <- args[2]
diag_path    <- args[3]
omnibus_path <- args[4]
chance_path  <- args[5]

df <- read.csv(data_path, stringsAsFactors = FALSE)
df$order        <- factor(df$order, levels = c("Human_first", "LLM_first"))
df$expertise    <- factor(df$expertise, levels = c("non-expert", "expert"))
df$pair_id      <- factor(df$pair_id)
df$score_diff_z <- as.numeric(df$score_diff_z)

cat("=== Replication check: observations per pair_id ===\\n")
pair_counts <- as.numeric(table(df$pair_id))
print(summary(pair_counts))
cat(sprintf("Pairs with only 1 rater: %d out of %d\\n",
            sum(pair_counts == 1), length(pair_counts)))

# ── Chance-level test ─────────────────────────────────────────────────────────
model_chance <- glmer(
  correct ~ 1 + (1 | pair_id),
  data = df, family = binomial(link = "logit"),
  control = glmerControl(optimizer = "bobyqa", optCtrl = list(maxfun = 2e5))
)
intercept_est      <- fixef(model_chance)["(Intercept)"]
intercept_se       <- sqrt(vcov(model_chance)["(Intercept)", "(Intercept)"])
z_val              <- intercept_est / intercept_se
p_val              <- 2 * pnorm(-abs(z_val))
predicted_accuracy <- plogis(intercept_est)

cat("=== Chance-level test ===\\n")
cat(sprintf("Model-implied accuracy: %.3f (chance = 0.500)\\n", predicted_accuracy))
cat(sprintf("Intercept = %.3f, SE = %.3f, z = %.3f, p = %.4f\\n",
            intercept_est, intercept_se, z_val, p_val))
write.csv(data.frame(
  predicted_accuracy = predicted_accuracy,
  intercept = intercept_est, se = intercept_se, z = z_val, p_value = p_val
), chance_path, row.names = FALSE)

# ── Full model ────────────────────────────────────────────────────────────────
model_full <- glmer(
  correct ~ author_gpt_vs_claude + author_gemini_vs_claude +
    order + expertise + score_diff_z + (1 | pair_id),
  data = df, family = binomial(link = "logit"),
  control = glmerControl(optimizer = "bobyqa", optCtrl = list(maxfun = 2e5))
)
cat("=== Full model: attribution accuracy ===\\n")
print(summary(model_full))

conv_warnings <- model_full@optinfo$conv$lme4$messages
cat("\\n=== Convergence ===\\n")
if (is.null(conv_warnings)) cat("No convergence warnings.\\n") else print(conv_warnings)

cat("\\n=== Overdispersion test (DHARMa) ===\\n")
sim_res   <- simulateResiduals(model_full, n = 1000)
disp_test <- testDispersion(sim_res, plot = FALSE)
print(disp_test)

cat("\\n=== Multicollinearity (VIF) ===\\n")
vif_result <- tryCatch(check_collinearity(model_full), error = function(e) NULL)
if (!is.null(vif_result)) print(vif_result)
max_vif <- if (!is.null(vif_result)) max(vif_result$VIF) else NA

cat("\\n=== Omnibus test: does Author matter? ===\\n")
model_no_author <- update(model_full, . ~ . - author_gpt_vs_claude - author_gemini_vs_claude)
omnibus_result  <- anova(model_full, model_no_author)
print(omnibus_result)
write.csv(data.frame(
  term = "Author (omnibus)",
  chisq = omnibus_result$Chisq[2],
  df = omnibus_result$"Df"[2],
  p_value = omnibus_result$"Pr(>Chisq)"[2]
), omnibus_path, row.names = FALSE)

tidy_fixed <- broom.mixed::tidy(model_full, effects = "fixed", conf.int = TRUE, conf.method = "Wald")
write.csv(tidy_fixed, fixed_path, row.names = FALSE)

write.csv(data.frame(
  overdispersion_detected = disp_test$p.value < 0.05,
  overdispersion_p_value  = disp_test$p.value,
  convergence_ok          = is.null(conv_warnings),
  max_vif                 = max_vif,
  vif_flag = ifelse(!is.na(max_vif) && max_vif > 5, "VIF > 5 (concern)", "OK")
), diag_path, row.names = FALSE)

cat("\\nDone.\\n")
'''

with open("analysis_attribution.R", "w") as f:
    f.write(r_script)

# ── Run R script ──────────────────────────────────────────────────────────────
result = subprocess.run(
    ["Rscript", "analysis_attribution.R", "attribution_data.csv",
     "glmm_attribution_fixed_effects.csv", "glmm_attribution_diagnostics.csv",
     "glmm_attribution_omnibus_author.csv", "glmm_attribution_chance_test.csv"],
    capture_output=True, text=True,
)
print(result.stdout)
print(result.stderr)
```

```python
source_omnibus = pd.read_csv("glmm_source_interaction_omnibus.csv")
source_simple = pd.read_csv("glmm_source_simple_effects.csv")
source_simple["item"] = source_simple["question_id"].map(QID_TO_ITEM)

def significance_label(p):
    if p < 0.001:
        return "Significant (p < .001)"
    elif p < 0.01:
        return "Significant (p < .01)"
    elif p < 0.05:
        return "Significant (p < .05)"
    else:
        return "Not significant"
        
def omnibus_row6b(label, omnibus_df):
    r = omnibus_df.iloc[0]
    return {
        "Term": label, "Estimate": "", "SE": "",
        "χ² / z": f"{r['chisq']:.2f}", "df": f"{int(r['df'])}",
        "p (FDR)": "<.001" if r["p_value"] < 0.001 else f"{r['p_value']:.3f}",
        "Sig.": significance_label(r["p_value"]),
    }

def simple_effect_row6b(item, contrast_row):
    p_adj = contrast_row["p.adj.fdr"]
    return {
        "Term": f"    {item}",
        "Estimate": f"{contrast_row['estimate']:.3f}",
        "SE": f"{contrast_row['SE']:.3f}",
        "χ² / z": f"{contrast_row['z.ratio']:.2f}",
        "df": "",
        "p (FDR)": "<.001" if p_adj < 0.001 else f"{p_adj:.3f}",
        "Sig.": significance_label(p_adj),
    }

blank_row6b = {"Term": "", "Estimate": "", "SE": "", "χ² / z": "", "df": "", "p (FDR)": "", "Sig.": ""}

rows6b = []
rows6b.append({**blank_row6b, "Term": "SOURCE × QUESTION INTERACTION"})
rows6b.append(omnibus_row6b("Omnibus test (does the Human/LLM gap vary by item?)", source_omnibus))
rows6b.append({**blank_row6b, "Term": "  Per-item simple effects (LLM − Human):"})
for _, r in source_simple.sort_values("question_id").iterrows():
    rows6b.append(simple_effect_row6b(r["item"], r))

table6b_df = pd.DataFrame(rows6b)

is_header6b = table6b_df["Term"] == "SOURCE × QUESTION INTERACTION"
is_subheader6b = table6b_df["Term"].str.strip() == "Per-item simple effects (LLM − Human):"
sig_labels = ["Significant (p < .001)", "Significant (p < .01)", "Significant (p < .05)"]
is_sig6b = table6b_df["Sig."].isin(sig_labels)
is_marginal6b = table6b_df["Sig."] == "Marginal (p < .10)"

def style_table6b(data):
    styles = pd.DataFrame("", index=data.index, columns=data.columns)
    styles.loc[is_header6b, :] = "font-weight: bold; background-color: #e5e5e5;"
    styles.loc[is_subheader6b, :] = "font-style: italic; color: #555555;"
    styles.loc[is_marginal6b, :] = "background-color: #fdf0e0;"
    styles.loc[is_sig6b, :] = "font-weight: bold; background-color: #dde3fb;"
    return styles

table6b = (
    table6b_df.style
    .apply(style_table6b, axis=None)
    .hide(axis="index")
    .set_caption(
        "Table 6b: Question-level interaction test — Source (Human/LLM). Tests whether the "
        "Human/LLM effect on pass rate varies depending on which specific TTCW question is "
        "being asked — a more specific question than Table 4/5's main-effects-only model can "
        "answer. Per-item p-values are FDR-corrected across all 14 items; this is what should "
        "be used to judge significance, not an uncorrected p-value. No rows are highlighted "
        "here — neither the omnibus test nor any individual item reaches significance or even "
        "marginal significance after correction, consistent with a clean null result."
    )
    .set_table_styles([
        {"selector": "thead th", "props": [("text-align", "left")]},
        {"selector": "caption", "props": [("text-align", "left")]},
        {"selector": "td", "props": [("text-align", "left")]},
    ])
)
table6b
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.cm as cm
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# ── Build per-rater item-level feature matrix ─────────────────────────────────
human_item_cols = [f"human_{item}" for item in TTCW_ITEMS]
llm_item_cols   = [f"llm_{item}"   for item in TTCW_ITEMS]

feature_cols = human_item_cols + llm_item_cols
rater_matrix = dataset_B[["responseID", "expertise"] + feature_cols].copy()
rater_matrix = rater_matrix.dropna(subset=feature_cols)

X         = rater_matrix[feature_cols].values
expertise = rater_matrix["expertise"].values

# ── Standardize ───────────────────────────────────────────────────────────────
scaler   = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ── PCA ───────────────────────────────────────────────────────────────────────
pca   = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

explained = pca.explained_variance_ratio_ * 100
print(f"PC1 explains {explained[0]:.1f}% of variance")
print(f"PC2 explains {explained[1]:.1f}% of variance")
print(f"Total:       {sum(explained):.1f}%")

# ── Silhouette scores to pick optimal k ───────────────────────────────────────
silhouette_scores = {}
for k in range(2, 8):
    km     = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_pca)
    silhouette_scores[k] = silhouette_score(X_pca, labels)

best_k = max(silhouette_scores, key=silhouette_scores.get)
print(f"\nSilhouette scores: { {k: round(v, 3) for k, v in silhouette_scores.items()} }")
print(f"Best k: {best_k}")

# ── Final k-means ─────────────────────────────────────────────────────────────
km_final      = KMeans(n_clusters=best_k, random_state=42, n_init=10)
cluster_labels = km_final.fit_predict(X_pca)

rater_matrix          = rater_matrix.copy()
rater_matrix["cluster"] = cluster_labels
rater_matrix["PC1"]     = X_pca[:, 0]
rater_matrix["PC2"]     = X_pca[:, 1]

# ── Dynamic cluster colors (works for any k) ──────────────────────────────────
cluster_colors = [cm.tab10(i) for i in range(best_k)]

# ── Plot 1 & 2: PCA by expertise and by cluster ───────────────────────────────
colors_exp = {"expert": "#534AB7", "non-expert": "#E8A838"}
labels_exp = {"expert": "Expert raters", "non-expert": "Non-expert raters"}

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Left: coloured by expertise
ax = axes[0]
for exp, group in rater_matrix.groupby("expertise"):
    ax.scatter(
        group["PC1"], group["PC2"],
        color=colors_exp[exp], label=labels_exp[exp],
        alpha=0.65, edgecolors="white", linewidths=0.5, s=70,
    )
ax.set_xlabel(f"PC1 ({explained[0]:.1f}% variance)", fontsize=11)
ax.set_ylabel(f"PC2 ({explained[1]:.1f}% variance)", fontsize=11)
ax.set_title("PCA — coloured by expertise", fontsize=13, fontweight="600")
ax.spines[["top", "right"]].set_visible(False)
handles_exp = [mpatches.Patch(color=colors_exp[e], label=labels_exp[e]) for e in colors_exp]
ax.legend(handles=handles_exp, frameon=False, fontsize=10)

# Right: coloured by k-means cluster
ax = axes[1]
for c in range(best_k):
    group = rater_matrix[rater_matrix["cluster"] == c]
    ax.scatter(
        group["PC1"], group["PC2"],
        color=cluster_colors[c], label=f"Cluster {c+1}",
        alpha=0.65, edgecolors="white", linewidths=0.5, s=70,
    )
ax.set_xlabel(f"PC1 ({explained[0]:.1f}% variance)", fontsize=11)
ax.set_ylabel(f"PC2 ({explained[1]:.1f}% variance)", fontsize=11)
ax.set_title(f"PCA — coloured by k-means cluster (k={best_k})", fontsize=13, fontweight="600")
ax.spines[["top", "right"]].set_visible(False)
ax.legend(frameon=False, fontsize=10)

fig.suptitle(
    "PCA of all 28 TTCW item scores (14 human + 14 LLM) per rater",
    fontsize=13, fontweight="600", y=1.01
)
plt.tight_layout()
plt.savefig("pca_cluster_raters.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Cluster composition table ─────────────────────────────────────────────────
print("\n=== Cluster composition by expertise ===")
comp          = rater_matrix.groupby(["cluster", "expertise"]).size().unstack(fill_value=0)
comp["total"] = comp.sum(axis=1)
print(comp)

# ── Mean TTCW scores per cluster ──────────────────────────────────────────────
print("\n=== Mean total TTCW score per cluster ===")
rater_matrix["human_total"] = rater_matrix[human_item_cols].sum(axis=1)
rater_matrix["llm_total"]   = rater_matrix[llm_item_cols].sum(axis=1)
cluster_means = rater_matrix.groupby("cluster")[["human_total", "llm_total"]].mean().round(2)
cluster_means.columns = ["Mean human score", "Mean LLM score"]
print(cluster_means)

# ── Silhouette plot ───────────────────────────────────────────────────────────
fig2, ax2 = plt.subplots(figsize=(6, 4))
ax2.plot(
    list(silhouette_scores.keys()), list(silhouette_scores.values()),
    marker="o", color="#534AB7", linewidth=2
)
ax2.axvline(best_k, color="#E8A838", linewidth=1.2, linestyle="--", label=f"Best k={best_k}")
ax2.set_xlabel("Number of clusters (k)", fontsize=11)
ax2.set_ylabel("Silhouette score", fontsize=11)
ax2.set_title("Silhouette scores by k", fontsize=12, fontweight="600")
ax2.legend(frameon=False, fontsize=10)
ax2.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.savefig("silhouette_scores.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
import pandas as pd
 
EXPERT_COLOR = "#A8D0E6"      # pastel blue
NONEXPERT_COLOR = "#F4B8B8"   # pastel coral/pink
 
def sig_color(p):
    if p < 0.05:
        return "#2c5f8a"   # significant
    elif p < 0.10:
        return "#d98f3f"   # marginal
    return "#aaaaaa"        # not significant
 
def wilson_ci(successes, n, z=1.96):
    if n == 0:
        return np.nan, np.nan
    p_hat = successes / n
    denom = 1 + z**2 / n
    center = (p_hat + z**2 / (2 * n)) / denom
    margin = (z * np.sqrt((p_hat * (1 - p_hat) + z**2 / (4 * n)) / n)) / denom
    return max(0.0, center - margin), min(1.0, center + margin)
```

```python
import matplotlib.colors as mcolors
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# ============================================================
# FONT — same minimalist stack as Figure 1
# ============================================================
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = ["Silka", "Helvetica Neue", "Arial", "DejaVu Sans"]

# ============================================================
# COLORS — two distinct pastel pink shades, same treatment as Figure 1
# ============================================================
EXPERT_COLOR = "#E8889B"      # deeper pastel pink/rose
NONEXPERT_COLOR = "#F6B8C4"   # lighter pastel pink

def lighten(hex_color, amount=0.45):
    """Blend a color toward white to get a lighter shade for whiskers."""
    rgb = np.array(mcolors.to_rgb(hex_color))
    white = np.array([1, 1, 1])
    return mcolors.to_hex(rgb + (white - rgb) * amount)

def wilson_ci(successes, n, confidence=0.95):
    """Wilson score confidence interval for a binomial proportion.
    More accurate than the normal approximation for small n or
    proportions near 0/1 -- standard choice for attribution-accuracy-style
    rates. Returns (ci_low, ci_high)."""
    if n == 0:
        return (np.nan, np.nan)
    z = norm.ppf(1 - (1 - confidence) / 2)
    p_hat = successes / n
    denom = 1 + z**2 / n
    center = (p_hat + z**2 / (2 * n)) / denom
    margin = (z / denom) * np.sqrt((p_hat * (1 - p_hat) / n) + (z**2 / (4 * n**2)))
    return (center - margin, center + margin)

attr_stats = (
    dataset_B.groupby("expertise")["attribution_correct"]
    .agg(successes="sum", n="count").reset_index()
)
attr_stats["rate"] = attr_stats["successes"] / attr_stats["n"]
ci3 = attr_stats.apply(lambda r: wilson_ci(r["successes"], r["n"]), axis=1, result_type="expand")
attr_stats[["ci_low", "ci_high"]] = ci3

fig3, ax3 = plt.subplots(figsize=(6, 5.5))
colors3 = [EXPERT_COLOR if e == "expert" else NONEXPERT_COLOR for e in attr_stats["expertise"]]
whisker_colors3 = [lighten(c, amount=0.45) for c in colors3]

x3 = np.arange(len(attr_stats))
rates3 = attr_stats["rate"].values * 100
yerr3 = np.vstack([
    rates3 - attr_stats["ci_low"].values * 100,
    attr_stats["ci_high"].values * 100 - rates3,
])

ax3.bar(x3, rates3, color=colors3, edgecolor="#888888", linewidth=0.6, width=0.55, zorder=2)

# Plot each bar's whisker individually so each gets its own lightened shade
for xi, ri, lo, hi, wc in zip(x3, rates3, yerr3[0], yerr3[1], whisker_colors3):
    ax3.errorbar(xi, ri, yerr=[[lo], [hi]], fmt="none", ecolor=wc, capsize=5, elinewidth=1.4, zorder=3)

ax3.axhline(50, color="black", linestyle="--", linewidth=1, label="Chance (50%)")

ax3.set_xticks(x3)
ax3.set_xticklabels(["Expert" if e == "expert" else "Non-expert" for e in attr_stats["expertise"]])
ax3.set_ylabel("% correctly identified AI-written story")
ax3.set_ylim(0, 100)
for xi, ri, ni in zip(x3, rates3, attr_stats["n"]):
    ax3.text(xi, ri + 3, f"n={ni}", ha="center", fontsize=9, color="#555")

ax3.set_title(
    "Figure 2. Attribution accuracy by expertise\n(bars = % correct, whiskers = 95% Wilson CI)",
    fontsize=12, fontweight="bold", loc="left",
)
ax3.legend(loc="upper right", frameon=False)
plt.tight_layout()
plt.savefig("fig2_attribution_by_expertise.png", dpi=200, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
import pandas as pd
from matplotlib.lines import Line2D

# ── Load coefficients from model output ───────────────────────────────────────
fixed = pd.read_csv("glmm_fixed_effects.csv")

def logit_to_prob(log_odds):
    return 1 / (1 + np.exp(-log_odds))

intercept = fixed.loc[fixed["term"] == "(Intercept)", "estimate"].values[0]

# ── Extract question_id rows ──────────────────────────────────────────────────
qid_rows = fixed[fixed["term"].str.startswith("question_id")].copy()
qid_rows["qid"] = qid_rows["term"].str.replace("question_id", "").astype(int)
qid_rows = qid_rows.sort_values("qid").reset_index(drop=True)

# ── Build full item list including reference (question_id 1 = Fluency1) ───────
TTCW_ITEMS_ORDERED = [
    "Fluency1", "Fluency2", "Fluency3", "Fluency4", "Fluency5",
    "Flexibility1", "Flexibility2", "Flexibility3",
    "Originality1", "Originality2", "Originality3",
    "Elaboration1", "Elaboration2", "Elaboration3",
]

# Reference item (Fluency1) has coef = 0, SE = 0
ref_row = pd.DataFrame([{
    "term": "question_id1", "qid": 1,
    "estimate": 0.0, "std.error": 0.0, "p.value": 1.0
}])
qid_rows = pd.concat([ref_row, qid_rows], ignore_index=True)
qid_rows["item"] = TTCW_ITEMS_ORDERED

# ── Compute probabilities ─────────────────────────────────────────────────────
qid_rows["prob"]    = qid_rows["estimate"].apply(lambda c: logit_to_prob(intercept + c))
qid_rows["prob_lo"] = qid_rows.apply(lambda r: logit_to_prob(intercept + r["estimate"] - 1.96 * r["std.error"]), axis=1)
qid_rows["prob_hi"] = qid_rows.apply(lambda r: logit_to_prob(intercept + r["estimate"] + 1.96 * r["std.error"]), axis=1)
qid_rows["sig"]     = qid_rows["p.value"] < 0.05

# ── Plot ──────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(7, 5))

for i, row in qid_rows.iterrows():
    color = "#E8945A" if row["sig"] else "#F5CFA8"
    ax.plot([row["prob_lo"], row["prob_hi"]], [i, i],
            color=color, linewidth=1.5, alpha=0.9)
    ax.scatter(row["prob"], i, color=color, s=50, zorder=3)

baseline = logit_to_prob(intercept)
ax.axvline(baseline, linestyle="--", color="#CCCCCC", linewidth=0.9, label=f"Fluency1 baseline ({baseline:.2f})")
ax.axvline(0.5, linestyle=":", color="#DDDDDD", linewidth=0.8)

ax.set_yticks(range(len(TTCW_ITEMS_ORDERED)))
ax.set_yticklabels(TTCW_ITEMS_ORDERED, fontsize=9, color="#555550")
ax.set_xlabel("Estimated pass probability", fontsize=9, color="#555550")
ax.set_title(
    "Estimated pass probability per TTCW item\n"
    "(darker = significantly harder than Fluency1, the reference item)",
    fontsize=10, color="#333330"
)
ax.set_xlim(0, 1)
ax.tick_params(colors="#888880")
ax.spines[["top", "right", "left", "bottom"]].set_visible(False)
ax.grid(axis="x", color="#EEEEEE", linewidth=0.8, zorder=0)

legend_elements = [
    Line2D([0], [0], marker="o", color="w", markerfacecolor="#E8945A",
           markersize=8, label="Significantly harder than Fluency1 (p < .05)"),
    Line2D([0], [0], marker="o", color="w", markerfacecolor="#F5CFA8",
           markersize=8, label="Not significantly different"),
    Line2D([0], [0], linestyle="--", color="#CCCCCC", linewidth=1,
           label=f"Fluency1 baseline ({baseline:.2f})"),
]
ax.legend(handles=legend_elements, fontsize=8, frameon=False)

plt.tight_layout()
plt.savefig("ttcw_pass_probability.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.lines as mlines
import numpy as np
import pandas as pd

# ── Pull rates from model_data ────────────────────────────────────────────────
rates = model_data.groupby(["dimension", "llm_author", "expertise"])["pass"].mean() * 100

def get_rate(dimension, author, expertise):
    try:
        return rates.loc[(dimension, author, expertise)]
    except KeyError:
        return np.nan

# ── Item order ────────────────────────────────────────────────────────────────
items = [
    ("Fluency",     "Fluency1",     "Narrative Pacing"),
    ("Fluency",     "Fluency2",     "Scene vs. Exposition"),
    ("Fluency",     "Fluency3",     "Literary Devices"),
    ("Fluency",     "Fluency4",     "Narrative Ending"),
    ("Fluency",     "Fluency5",     "Understandability & Coherence"),
    ("Flexibility", "Flexibility1", "Perspective & Voice Flexibility"),
    ("Flexibility", "Flexibility2", "Emotional Flexibility"),
    ("Flexibility", "Flexibility3", "Structural Flexibility"),
    ("Originality", "Originality1", "Originality in Theme & Content"),
    ("Originality", "Originality2", "Originality in Thought"),
    ("Originality", "Originality3", "Originality in Form & Structure"),
    ("Elaboration", "Elaboration1", "World Building & Setting"),
    ("Elaboration", "Elaboration2", "Character Development"),
    ("Elaboration", "Elaboration3", "Rhetorical Complexity"),
]

questions = [x[2] for x in items]
dims      = [x[1] for x in items]
n         = len(questions)

models  = ["GPT", "Gemini", "Claude"]
authors = {"GPT": "gpt", "Gemini": "gemini", "Claude": "claude"}

EXPERT_COLOR    = "#E05C5C"   # red
NONEXPERT_COLOR = "#4CAF50"   # green

# ── Compute deviations from human baseline ────────────────────────────────────
def build_deviations(expertise_filter=None):
    devs = {m: [] for m in models}
    for dim in dims:
        if expertise_filter:
            human_df = model_data[
                (model_data["dimension"] == dim) &
                (model_data["source"] == "Human") &
                (model_data["expertise"] == expertise_filter)
            ]
        else:
            human_df = model_data[
                (model_data["dimension"] == dim) &
                (model_data["source"] == "Human")
            ]
        human_rate = human_df["pass"].mean() * 100

        for m in models:
            llm_df = model_data[
                (model_data["dimension"] == dim) &
                (model_data["llm_author"] == authors[m])
            ]
            if expertise_filter:
                llm_df = llm_df[llm_df["expertise"] == expertise_filter]
            llm_rate = llm_df["pass"].mean() * 100
            devs[m].append(llm_rate - human_rate)
    return devs

dev_total  = build_deviations()
dev_expert = build_deviations("expert")
dev_nonexp = build_deviations("non-expert")

# ── Compute dynamic MAX ───────────────────────────────────────────────────────
all_vals = []
for m in models:
    all_vals.extend(dev_total[m])
    all_vals.extend(dev_expert[m])
    all_vals.extend(dev_nonexp[m])
all_vals = [v for v in all_vals if not np.isnan(v)]
MAX = max(abs(min(all_vals)), abs(max(all_vals))) + 3
MAX = max(MAX, 17)

# ── Plot setup ────────────────────────────────────────────────────────────────
colors  = {"GPT": "#2a78d6", "Gemini": "#1baf7a", "Claude": "#eda100"}
offsets = {"GPT": -0.22,     "Gemini": 0.0,       "Claude":  0.22}

row_heights = []
for i in range(n):
    row_heights.append(1)
    if i < n - 1:
        row_heights.append(0.12)

fig, axes = plt.subplots(
    len(row_heights), 1,
    figsize=(8, n * 0.65 + 0.6),
    gridspec_kw={"height_ratios": row_heights, "hspace": 0}
)
fig.patch.set_facecolor("white")

data_axes = []
for idx, ax in enumerate(axes):
    if idx % 2 == 0:
        data_axes.append(ax)
    else:
        ax.axis("off")

# ── Draw rows ─────────────────────────────────────────────────────────────────
for i, (ax, q) in enumerate(zip(data_axes, questions)):
    ax.set_facecolor("#f8f8f8" if i % 2 == 0 else "white")
    ax.set_xlim(-MAX, MAX)
    ax.set_ylim(-0.6, 0.6)
    ax.set_yticks([])
    ax.spines[["top", "bottom", "left", "right"]].set_visible(False)

    ax.axvline(0, color="#c3c2b7", linewidth=0.9, zorder=1)
    for tick in [-20, -15, -10, -5, 5, 10, 15, 20]:
        if abs(tick) <= MAX:
            ax.axvline(tick, color="#e8e8e8", linewidth=0.7, zorder=0)

    for model in models:
        y     = offsets[model]
        color = colors[model]

        val_total  = np.clip(dev_total[model][i],  -MAX + 1, MAX - 1)
        val_expert = np.clip(dev_expert[model][i], -MAX + 1, MAX - 1)
        val_nonexp = np.clip(dev_nonexp[model][i], -MAX + 1, MAX - 1)

        lo = min(val_expert, val_nonexp)
        hi = max(val_expert, val_nonexp)

        # whisker line in model color
        ax.plot([lo, hi], [y, y],
                color=color, linewidth=1.0, alpha=0.4,
                zorder=2, solid_capstyle="butt")

        # expert point — red
        ax.scatter(val_expert, y, color=EXPERT_COLOR,
                   s=28, zorder=3, clip_on=False,
                   edgecolor="white", linewidth=0.4)

        # non-expert point — green
        ax.scatter(val_nonexp, y, color=NONEXPERT_COLOR,
                   s=28, zorder=3, clip_on=False,
                   edgecolor="white", linewidth=0.4)

        # main total dot — model color, larger, on top
        ax.scatter(val_total, y, color=color, s=50,
                   zorder=4, clip_on=False,
                   edgecolor="white", linewidth=0.5)

    ax.text(-MAX - 0.5, 0, q, ha="right", va="center",
            fontsize=8.5, color="#333333", transform=ax.transData)

    if i == n - 1:
        tick_vals = [t for t in [-20, -15, -10, -5, 0, 5, 10, 15, 20]
                     if abs(t) <= MAX]
        ax.set_xticks(tick_vals)
        ax.xaxis.set_major_formatter(
            plt.FuncFormatter(lambda x, _: f"{x:+.0f}pp")
        )
        ax.tick_params(axis="x", labelsize=8, colors="#888")
        ax.spines["bottom"].set_visible(True)
        ax.spines["bottom"].set_color("#c3c2b7")
    else:
        ax.set_xticks([])

# ── Legend ────────────────────────────────────────────────────────────────────
handles = [mpatches.Patch(color=colors[m], label=m) for m in models]
handles += [
    mlines.Line2D([], [], marker="o", color=EXPERT_COLOR,
                  linestyle="None", markersize=6, label="Expert raters"),
    mlines.Line2D([], [], marker="o", color=NONEXPERT_COLOR,
                  linestyle="None", markersize=6, label="Non-expert raters"),
    mlines.Line2D([], [], marker="o", color="grey",
                  linestyle="None", markersize=7, label="Total (all raters)"),
]
fig.legend(handles=handles, fontsize=8.5, frameon=False,
           loc="upper right", bbox_to_anchor=(0.98, 0.99), ncol=3)

plt.subplots_adjust(left=0.35, right=0.97, top=0.94, bottom=0.04)
plt.savefig("fig_llm_deviation_expertise.png", dpi=200, bbox_inches="tight")
plt.show()
print("Saved to fig_llm_deviation_expertise.png")
```
